## Task 2.3

Inserting $u(t, x, y) = e^{i(k_x x + k_y y - \omega t)}$ in the wave equation $\frac{\partial^2 u}{\partial t^2} = c^2 \nabla^2 u$. We have the derivatives:

$$
\frac{\partial^2 u}{\partial t^2} = -\omega^2 u, \quad \frac{\partial^2 u}{\partial x^2} = -k_x^2 u, \quad \frac{\partial^2 u}{\partial y^2} = -k_y^2 u
$$

so we get

$$
-\omega^2 u = -c^2 (k_x^2 + k_y^2) u
$$

So $u$ satisfeis the wave equation when $\omega$ satisfies the dispersion relation
$$
\omega = c \sqrt{k_x^2 + k_y^2}
$$

## Task 2.4

$k_x = k_y = k$. Since $x_i = ih$ and $y_j = jh$ we can rewrite $u^n_{ij} = e^{i(kh(i+j) - \tilde{\omega} n \Delta t)}$ as:

$$
u^n_{ij} = A^n e^{ik(x_i + y_j)}
$$

where $A = e^{-i \tilde{\omega} \Delta t}$. Inserting $u^n_{ij}$ into the discretized wave equation and we get:

$$
\frac{\left(A^{n+1} - 2A^n + A^{n-1}\right) e^{ik(x_i + y_j)}}{\Delta t^2} = c^2 A^n \left( \frac{e^{ik(x_i + h + y_j)} - 2e^{ik(x_i + y_j)} + e^{ik(x_i - h + y_j)}}{h^2} + \frac{e^{ik(x_i + y_j + h)} - 2e^{ik(x_i + y_j)} + e^{ik(x_i + y_j - h)}}{h^2} \right)
$$

Dividing by $A^n e^{ik(x_i + y_j)}$, multiply by $\Delta t^2$ and use $C = c \Delta t / h$ and get:

$$
A - 2 + A^{-1} = 2C^2 \left(e^{ikh} - 2 + e^{-ikh}\right)
$$

Since $e^{ix} + e^{-ix} = 2 \cos x$ we get:

$$
A + A^{-1} = 2 + 4C^2 (\cos kh - 1)
$$

Since $A + A^{-1} = e^{-i \tilde{\omega} \Delta t} + e^{i \tilde{\omega} \Delta t} = 2 \cos (\tilde{\omega} \Delta t)$, this is

$$
\cos (\tilde{\omega} \Delta t) = 1 + 2C^2 (\cos kh - 1)
$$

Using $C = 1/\sqrt{2}$ we get $\cos (\tilde{\omega} \Delta t) = \cos kh$, so that $\tilde{\omega} \Delta t = kh$. With $\Delta t = Ch/c = h/(\sqrt{2} c)$

$$
\tilde{\omega} = \frac{kh}{\Delta t} = \sqrt{2} c k = c \sqrt{k^2 + k^2} = \omega
$$

```python
import sys
sys.path.insert(0, "..")
import numpy as np
import matplotlib.pyplot as plt
from Wave2D import Wave2D_Neumann

sol = Wave2D_Neumann()
data = sol(40, 40, cfl=1/np.sqrt(2), mx=2, my=2, store_data=2)
xij, yij = sol.xij, sol.yij
```

```python
%%capture
import matplotlib.animation as animation

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
frames = []
for n, val in data.items():
    frame = ax.plot_wireframe(xij, yij, val, rstride=2, cstride=2);
    frames.append([frame])

ani = animation.ArtistAnimation(fig, frames, interval=400, blit=True,
                                repeat_delay=1000)
ani.save('neumannwave.gif', writer='pillow', fps=5)
```
